```python
# 1. Fill missing Prices (Total / Qty)
df['Price Per Unit'] = df['Price Per Unit'].fillna(df['Total Spent'] / df['Quantity'])

```

```python
df.isnull().sum()
```

```python
# 2. Fill missing Quantities (Total / Price)
df['qty'] = df['qty'].fillna(df['total'] / df['price'])

```

```python
# 3. Fill missing Totals (Qty * Price)
df['total'] = df['total'].fillna(df['qty'] * df['price'])
```
