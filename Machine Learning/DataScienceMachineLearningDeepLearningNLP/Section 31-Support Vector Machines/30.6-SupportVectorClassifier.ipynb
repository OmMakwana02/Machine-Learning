```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
from sklearn.datasets import make_classification
```

```python
X, y = make_classification(n_samples=1000, n_features=2, n_classes=2, n_clusters_per_class=1, n_redundant=0)
```

```python
X, y
```

```python
df = pd.DataFrame(X, columns=['X1', 'X2'])
df['Label'] = y
sns.scatterplot(x=pd.DataFrame(X)[0], y=pd.DataFrame(X)[1], hue=y, data=df)
plt.show()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
from sklearn.svm import SVC

svc = SVC(kernel = 'linear')
```

```python
svc.fit(X_train, y_train)
```

```python
y_pred = svc.predict(X_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```

```python
X, y = make_classification(n_samples=1000, n_features=2, n_classes=2, n_clusters_per_class=2, n_redundant=0)

df = pd.DataFrame(X, columns=['X1', 'X2'])
df['Label'] = y
sns.scatterplot(x=pd.DataFrame(X)[0], y=pd.DataFrame(X)[1], hue=y, data=df)
plt.show()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
from sklearn.svm import SVC

svc = SVC(kernel = 'linear')

svc.fit(X_train, y_train)
```

```python
y_pred = svc.predict(X_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print(classification_report(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
```

```python
rbf = SVC(kernel = 'rbf')
```

```python
rbf.fit(X_train, y_train)
```

```python
y_pred_rbf =rbf.predict(X_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print(classification_report(y_test, y_pred_rbf))
print(confusion_matrix(y_test, y_pred_rbf))
```

```python
poly = SVC(kernel = 'poly')
```

```python
poly.fit(X_train, y_train)
```

```python
y_pred_poly = poly.predict(X_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print(classification_report(y_test, y_pred_poly))
print(confusion_matrix(y_test, y_pred_poly))
```

```python
sig = SVC(kernel='sigmoid')
sig.fit(X_train, y_train)
```

```python
y_pred_sig = sig.predict(X_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print(classification_report(y_test, y_pred_sig))
print(confusion_matrix(y_test, y_pred_sig))
```

As you can see we have used different different kernels to train the model, and we will be selecting the kernel whose accuracy is best, in our case, rbf.
